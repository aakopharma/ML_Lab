## **Write a Program to find the mean, median, standard deviation and mode using user-defined functions**.

```python
import math
def mean(data):
    n = len(data)
    result = 0
    for x in data:
        result = result + x

    average = result/n
    return average
```

```python
def median(data):
    data.sort()
    if(len(data)%2!=0):
        i = len(data)//2
        return data[i]
    index = len(data)//2
    # return mean([data[index-1],data[index]])
    ans = data[index-1]+data[index]
    return ans/2
```

```python
def standard_deviation(data):
    mu = mean(data)
    res = 0
    for x in data:
        res = res + (x-mu)**2
    res = res/len(data)
    return math.sqrt(res)
standard_deviation([8,5,2,10,15])
```
