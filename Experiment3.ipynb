```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('revenue_prediction.csv')
df
```

```python
# Id and name of the place are not important to predict revenue, so we are going to remove them 
# df = df.drop('Id',axis=1)
df.drop('Id',axis=1,inplace=True)  #It will update the changes in actual df.
```

```python
df.drop(columns=['Id','Name'],axis=1,inplace=True)
```

```python
df
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
df['City'].unique()
```

```python
le = LabelEncoder()

df['City'] = le.fit_transform(df['City'])
```

```python
for x in df:
    if(x=='City'):
        break
    df[x] = le.fit_transform(df[x])
    
```

```python
df
```

```python
X = df.iloc[:,3:-1]
X
```

```python
y = df['Revenue']
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(X,y,train_size=0.8,random_state=0)
x_train
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(x_train,y_train)  #Passing training data to the model
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn.metrics import *
```

```python
r2_score(y_test,y_pred)
```

```python
rmse = np.sqrt(mean_squared_error(y_test,y_pred))
rmse
```

```python
mean_squared_error(y_test,y_pred)
```
